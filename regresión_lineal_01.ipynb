## Descripción del proceso de regresión

La **regresión lineal simple** permite representar mediante una recta la relación entre dos variables numéricas. En este notebook se estudiara el caso de un sensor de temperatura:

- $x$: temperatura indicada por un instrumento de referencia;
- $y$: lectura entregada por el sensor que se desea evaluar.

El objetivo es construir el modelo

$$\hat y=b_0+b_1x,$$

donde $\hat y$ es la lectura estimada, $b_0$ es el intercepto y $b_1$ es la pendiente. Los coeficientes se obtendrán por **mínimos cuadrados**, método que selecciona la recta que minimiza la suma de los residuos al cuadrado.

El proceso comprende organizar y revisar los datos, visualizar su relación, calcular los coeficientes, producir predicciones, estudiar los residuos y evaluar el ajuste mediante RMSE y $R^2$. Antes de programar cada celda, lea toda la explicación, identifique las operaciones requeridas y use los nombres de variables indicados.

7. Obtener la pendiente:

   $$b_1=\frac{\sum(x_i-\bar{x})(y_i-\bar{y})}{\sum(x_i-\bar{x})^2}.$$

8. Obtener el intercepto: $b_0=\bar{y}-b_1\bar{x}$.
9. Calcular predicciones $\hat y_i$ y residuos $e_i=y_i-\hat y_i$.
